```python
import kagglehub
import pandas as pd
import os
import glob

# حمّل الداتا واخد المسار الصح تلقائي
path = kagglehub.dataset_download("dhoogla/unswnb15")

# دوّر على كل الملفات
files = glob.glob(os.path.join(path, "**", "*"), recursive=True)

# اختار ملفات CSV أو Parquet
data_files = [
    f for f in files
    if os.path.isfile(f)
    and f.lower().endswith((".csv", ".parquet"))
]

print("Available data files:")
for i, file in enumerate(data_files):
    print(i, os.path.basename(file))

# حمّل أول ملف متاح
if data_files:
    file_path = data_files[0]

    if file_path.lower().endswith(".csv"):
        df = pd.read_csv(file_path)
    else:
        df = pd.read_parquet(file_path)

    # اعرض أول 100 صف
    display(df.head(100))

    # اعرض أبعاد الداتا
    print("Dataset shape:", df.shape)
else:
    print("No CSV or Parquet files were found.")
```

```python
for col in df.select_dtypes(include='number').columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, len(outliers))
```
