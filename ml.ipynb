```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Use your transformed dataset
df = df.copy()

print("Dataset shape:", df.shape)

# 2. Target column
target = "label"

if target not in df.columns:
    raise ValueError("Label column not found!")

# 3. Select numeric feature columns
numeric_df = df.select_dtypes(include=np.number)

feature_cols = [
    col for col in numeric_df.columns
    if col not in [target, "attack_cat"]
]

X = numeric_df[feature_cols]

# 4. Correlation matrix
corr_matrix = X.corr(method="pearson")

# 5. Heatmap with correlation numbers
plt.figure(figsize=(20, 16))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    annot_kws={"size": 8},
    linewidths=0.5
)

plt.title("Correlation Matrix - UNSW-NB15", fontsize=16)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# 6. Find highly correlated pairs
threshold = 0.90

upper = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

high_corr = (
    upper.stack()
    .loc[lambda s: s.abs() > threshold]
    .sort_values(key=abs, ascending=False)
)

print("\nHighly correlated pairs:")
print(
    high_corr
    if not high_corr.empty
    else "No pairs above the threshold."
)

# 7. Correlation with target
target_corr = numeric_df[feature_cols].corrwith(
    numeric_df[target]
).abs()

# 8. Select columns to remove
to_drop = set()

for (col1, col2), value in high_corr.items():

    if col1 in to_drop or col2 in to_drop:
        continue

    score1 = target_corr.get(col1, 0)
    score2 = target_corr.get(col2, 0)

    if pd.isna(score1):
        score1 = 0

    if pd.isna(score2):
        score2 = 0

    # Keep the feature more correlated with label
    if score1 >= score2:
        to_drop.add(col2)
    else:
        to_drop.add(col1)

print("\nSuggested columns to remove:")
print(sorted(to_drop))

# 9. Create reduced dataset
reduced_df = df.drop(
    columns=list(to_drop)
)

print("\nBefore:", df.shape)
print("After:", reduced_df.shape)

# 10. Save reduced dataset
reduced_df.to_csv(
    "UNSW_NB15_reduced.csv",
    index=False
)

print("\nReduced dataset saved successfully!")
```
