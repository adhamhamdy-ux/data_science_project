```python
import kagglehub
import pandas as pd
import os
import glob

# حمّل الداتا واخد المسار الصح تلقائي
path = kagglehub.dataset_download("dhoogla/unswnb15")

# دوّر على كل الملفات
files = glob.glob(os.path.join(path, "**", "*"), recursive=True)

# اختار ملفات CSV أو Parquet
data_files = [
    f for f in files
    if os.path.isfile(f)
    and f.lower().endswith((".csv", ".parquet"))
]

print("Available data files:")
for i, file in enumerate(data_files):
    print(i, os.path.basename(file))

# حمّل أول ملف متاح
if data_files:
    file_path = data_files[0]

    if file_path.lower().endswith(".csv"):
        df = pd.read_csv(file_path)
    else:
        df = pd.read_parquet(file_path)

    # اعرض أول 100 صف
    display(df.head(100))

    # اعرض أبعاد الداتا
    print("Dataset shape:", df.shape)
else:
    print("No CSV or Parquet files were found.")
```

```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Use your transformed dataset
# Replace df_transformed with your actual DataFrame name
df = df.copy()

print("Dataset shape:", df.shape)

# Target column
target = "label"

# Check that target exists
if target not in df.columns:
    print("Label column not found. Available columns:")
    print(df.columns.tolist())
else:
    # Select numeric feature columns, excluding the target
    numeric_df = df.select_dtypes(include=np.number)

    feature_cols = [
        col for col in numeric_df.columns
        if col not in [target, "attack_cat"]
    ]

    X = numeric_df[feature_cols]

    # Correlation matrix
    corr_matrix = X.corr(method="pearson")

    # Heatmap
    plt.figure(figsize=(20, 16))
    sns.heatmap(
        corr_matrix,
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1
    )
    plt.title("Correlation Matrix - UNSW-NB15")
    plt.tight_layout()
    plt.show()

    # Find highly correlated pairs
    threshold = 0.90

    upper = corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )

    high_corr = (
        upper.stack()
        .loc[lambda s: s.abs() > threshold]
        .sort_values(key=abs, ascending=False)
    )

    print("\nHighly correlated pairs:")
    print(high_corr if not high_corr.empty
          else "No pairs above the threshold.")

    # Correlation with binary target
    if target in numeric_df.columns:
        target_corr = numeric_df[feature_cols].corrwith(
            numeric_df[target]
        ).abs()

        to_drop = set()

        for (col1, col2), value in high_corr.items():
            if col1 in to_drop or col2 in to_drop:
                continue

            score1 = target_corr.get(col1, 0)
            score2 = target_corr.get(col2, 0)

            if pd.isna(score1):
                score1 = 0
            if pd.isna(score2):
                score2 = 0

            # Keep the feature more correlated with label
            if score1 >= score2:
                to_drop.add(col2)
            else:
                to_drop.add(col1)

        print("\nSuggested columns to remove:")
        print(sorted(to_drop))

        # Create reduced dataset
        reduced_df = df.drop(
            columns=list(to_drop)
        )

        print("\nBefore:", df.shape)
        print("After:", reduced_df.shape)

        # Save the result
        reduced_df.to_csv(
            "UNSW_NB15_reduced.csv",
            index=False
        )
```
