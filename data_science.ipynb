```python
import kagglehub
import pandas as pd
import os
import glob

files = glob.glob(
    os.path.join(
        r"C:\Users\hp\.cache\kagglehub\datasets\dhoogla\unswnb15\versions\5",
        "**",
        "*"
    ),
    recursive=True
)

# Select a CSV or Parquet file
data_files = [
    f for f in files
    if os.path.isfile(f)
    and f.lower().endswith((".csv", ".parquet"))
]

print("Available data files:")
for i, file in enumerate(data_files):
    print(i, os.path.basename(file))

# Load the first available data file
if data_files:
    file_path = data_files[0]

    if file_path.lower().endswith(".csv"):
        df = pd.read_csv(file_path)
    else:
        df = pd.read_parquet(file_path)

    # Display the first 100 rows
    display(df.head(100))

    # Display dataset dimensions
    print("Dataset shape:", df.shape)
else:
    print("No CSV or Parquet files were found.")
```
